```python
from datetime import datetime

# Original list in row-major order
date_strings = [
    "8 Jun 2000", "1 Sep 1997", "30 Sep 1999", "1 Oct 2008", "12 Sep 1998", "15 Jun 2003",
    "17 Sep 1996", "3 Sep 1997", "8 Oct 1995", "28 Sep 1996", "24 Jul 2008", "18 Aug 1995",
    "17 Sep 2005", "6 Aug 2003", "2 Sep 2002", "1 Oct 2007", "1 Oct 2007", "18 Sep 2003",
    "5 Sep 2010", "21 Aug 1995", "5 Oct 2002", "2 Oct 2009", "27 Sep 2016", "5 Sep 2010",
    "20 Sep 2005", "3 Oct 1994", "2 Oct 2002", "23 Aug 1998", "29 Sep 2008", "14 Sep 2003"
]

# Reshape into 5 rows × 6 columns
rows, cols = 5, 6
grid = [date_strings[i * cols:(i + 1) * cols] for i in range(rows)]

# Reorder column-wise
column_major_order = [grid[row][col] for col in range(cols) for row in range(rows)]

# Convert to datetime objects
def parse_date(date_str):
    try:
        return datetime.strptime(date_str, "%d %b %Y")
    except ValueError:
        return datetime.strptime(date_str, "%d %B %Y")

indts = [parse_date(date_str) for date_str in column_major_order]
indts = [i+relativedelta(hours=12) for i in indts]
```
